```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35], 
    'City': ['NYC', 'LA', 'Chicago']
}
df = pd.DataFrame(data)
```

```python
display(df)

# Boolean indexing
young_people = df[df['Age'] < 30]
display(young_people)

# Multiple conditions
ny_young = df[(df['Age'] < 30) & 
              (df['City'] == 'NYC')]
display(ny_young)
    
# Using query method
result = df.query('Age > 25 and City == "LA"')
display(result)
```

```python
# Group by city
city_groups = df.groupby('City')

# Aggregate functions
avg_age_by_city = city_groups['Age'].mean()
display(avg_age_by_city)

# Multiple aggregations
stats = city_groups.agg({
    'Age': ['mean', 'max', 'min']
})
display(stats)
```

```python
# Sort by values
sorted_df = df.sort_values('Age', 
                          ascending=False)
display(sorted_df)

# Sort by multiple columns
multi_sort = df.sort_values(['City', 'Age'])
display(multi_sort)

# Rank values
df['Age_Rank'] = df['Age'].rank()
display(df)

sorted_df['Age_Rank'] = sorted_df['Age'].rank()
display(sorted_df)
```
