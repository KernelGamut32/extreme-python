# Demo: Comparing `.loc[]` and `.iloc[]` in Pandas

```python
# Step 1: Import libraries
import pandas as pd

# Step 2: Create a simple DataFrame
df = pd.DataFrame({
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'Population': [8419600, 3980400, 2716000, 2328000, 1690000],
    'Region': ['East', 'West', 'Midwest', 'South', 'West']
}, index=['NY', 'LA', 'CH', 'HO', 'PH'])

# Display the DataFrame
print("=== Original DataFrame ===")
display(df)
```

```python
display(df)

# Select one row by label
print("Select row with index label 'CH':")
display(df.loc['CH'])

# Select multiple rows by label
print("Select rows 'NY' and 'HO':")
display(df.loc[['NY', 'HO']])

# Select a range of rows by label (note: inclusive!)
print("Select rows from 'LA' to 'HO':")
display(df.loc['LA':'HO'])
```

```python
display(df)

# Select one row by integer position
print("Select the first row (position 0):")
display(df.iloc[0])

# Select multiple rows by position
print("Select rows at positions 1 and 3:")
display(df.iloc[[1, 3]])

# Select a range of rows by position (note: exclusive!)
print("Select rows at positions 1 through 3:")
display(df.iloc[1:3])
```

**NOTE**: `.iloc[1:3]` includes rows 1 and 2, but not 3 (different from label-based selection using `loc[]`)

### Selecting Rows and Columns

```python
display(df)

# loc: uses row & column labels
print("Using .loc: select rows 'NY' to 'CH' and columns 'City' & 'Population'")
display(df.loc['NY':'CH', ['City', 'Population']])

# iloc: uses integer positions
print("Using .iloc: select first 3 rows and first 2 columns")
display(df.iloc[0:3, 0:2])
```

### Boolean Filtering with `loc[]`

```python
display(df)

# Boolean condition: cities with population > 3 million
print("Cities with population > 3 million (using loc):")
display(df.loc[df['Population'] > 3_000_000])
```

| Feature                   | `.loc[]`    | `.iloc[]`              |
| ------------------------- | ----------- | ---------------------- |
| Indexing type             | Label-based | Integer position-based |
| End value in slice        | Inclusive   | Exclusive              |
| Boolean masking           | ✅ Supported | ✅ Supported            |
| Accepts integer positions | ❌           | ✅                      |
| Accepts labels            | ✅           | ❌                      |
