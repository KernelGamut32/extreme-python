```python
import pandas as pd

# Create sample wide data
df_wide = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet'],
    'Jan': [100, 200, 150],
    'Feb': [120, 180, 160],
    'Mar': [110, 220, 140]
})

# Convert to tall format
df_tall = df_wide.melt(
    id_vars=['Product'],
    var_name='Month',
    value_name='Sales'
)

print(df_wide.head())
print(df_tall.head())
```

```python
# MultiIndex DataFrame
df_multi = pd.DataFrame({
    ('Sales', 'Q1'): [100, 90],
    ('Sales', 'Q2'): [120, 100],
    ('Profit', 'Q1'): [20, 15],
    ('Profit', 'Q2'): [25, 18]
})

# Stack to tall format
df_stacked = df_multi.stack(future_stack=True)

print(df_multi)
print(df_stacked)
```

```python
# Starting with tall format
df_tall = pd.DataFrame({
    'Store': ['A', 'A', 'B', 'B'],
    'Product': ['Laptop', 'Phone', 'Laptop', 'Phone'],
    'Revenue': [5000, 3000, 4500, 3200]
})

# Pivot to wide format
df_pivot = df_tall.pivot(
    index='Store',
    columns='Product',
    values='Revenue'
)

print(df_tall)
print(df_pivot)
```
