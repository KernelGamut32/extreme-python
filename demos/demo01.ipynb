```python
# From list
s1 = pd.Series([1, 2, 3, 4])

# From dictionary  
s2 = pd.Series({'a': 1, 'b': 2, 'c': 3})

# With custom index
s3 = pd.Series([10, 20, 30], 
               index=['x', 'y', 'z'])
```

```python
s1, s2, s3
```

```python
# By label
print(s3['x'])  # 10

# By position
print(s3.iloc[0])  # 10

# Multiple values
print(s3[['x', 'z']])
```

```python
display(s3)

# Mathematical operations
result = s3 * 2
display(result)
display(s3)

# Boolean indexing
filtered = s3[s3 > 15]
display(filtered)
display(s3)

# Statistical methods
print(s3.mean(), s3.std())
```
